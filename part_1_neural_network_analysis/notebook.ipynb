#### part 1 changing dtypes and standardizing the data

```python
df.head()
```

```python
df.columns
```

```python
# maping str to int
df['region'] = df['region'].map({'South': 0, 'West': 1, 'Central': 2, 'North': 3, 'East': 4})
df['plan_type'] = df['plan_type'].map({'Basic': 0, 'Standard': 1, 'Premium': 2, 'Enterprise':3})
df['contract_type'] = df['contract_type'].map({'Month-to-month': 0, 'One-year': 1, 'Two-year': 2})
df['payment_method'] = df['payment_method'].map({'Debit Card': 0, 'Wallet': 1, 'Credit Card': 2, 'Net Banking': 3, 'UPI': 4})
df.head()
```

```python
df.info()
```

```python
df.head()
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
columns = ['tenure_months', 'monthly_charges', 'total_charges']
df[columns] = scaler.fit_transform(df[columns])
df.head()
```

```python
from sklearn.model_selection import train_test_split
x = df.drop('churn', axis=1)
y = df['churn']
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
print(x_train.shape, y_train.shape)
```
